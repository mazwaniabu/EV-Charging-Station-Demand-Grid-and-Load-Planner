# EV & Hybrid Adoption and Charging Infrastructure Analysis

This notebook performs exploratory data analysis (EDA) on the cleaned EV/Hybrid registration datasets and charging station datasets for Malaysia.

```python
import os
import pandas as pd
import plotly.express as px

# File Paths
CARS_PATH = os.path.join("..", "data", "processed", "cleaned_cars_2026.csv")
STATIONS_PATH = os.path.join("..", "data", "processed", "cleaned_charging_station.csv")
```

## 1. Load and Inspect Cleaned Data

```python
cars_df = pd.read_csv(CARS_PATH)
stations_df = pd.read_csv(STATIONS_PATH)

print(f"Cleaned Cars Dataset: {cars_df.shape[0]} records, {cars_df.shape[1]} columns")
print(f"Cleaned Charging Stations Dataset: {stations_df.shape[0]} records, {stations_df.shape[1]} columns")
```

## 2. EV & Hybrid Adoption by State

```python
# Calculate distribution
cars_by_state = cars_df.groupby('state').size().reset_index(name='vehicle_count')
cars_by_state = cars_by_state.sort_values(by='vehicle_count', ascending=False).reset_index(drop=True)

print("--- Top States for EV/Hybrid Adoption ---")
print(cars_by_state.head(10))
```

```python
# Plot adoption chart
fig_cars = px.bar(
    cars_by_state,
    x='vehicle_count',
    y='state',
    orientation='h',
    title='EV & Hybrid Adoption by State in Malaysia (2026)',
    labels={'vehicle_count': 'Number of Registered Vehicles', 'state': 'State'},
    color='vehicle_count',
    color_continuous_scale=px.colors.sequential.Viridis
)
fig_cars.update_layout(yaxis={'categoryorder':'total ascending'}, showlegend=False)
fig_cars.show()
```

## 3. Charging Infrastructure Distribution by State

```python
# Calculate charging stations distribution
stations_by_state = stations_df.groupby('state_province').size().reset_index(name='station_count')
stations_by_state = stations_by_state.sort_values(by='station_count', ascending=False).reset_index(drop=True)

print("--- Top States for Charging Infrastructure ---")
print(stations_by_state.head(10))
```

```python
# Plot charging stations chart
fig_stations = px.bar(
    stations_by_state,
    x='station_count',
    y='state_province',
    orientation='h',
    title='Charging Station Count by State in Malaysia',
    labels={'station_count': 'Number of Charging Stations', 'state_province': 'State/Province'},
    color='station_count',
    color_continuous_scale=px.colors.sequential.Cividis
)
fig_stations.update_layout(yaxis={'categoryorder':'total ascending'}, showlegend=False)
fig_stations.show()
```
